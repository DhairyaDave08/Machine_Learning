# **Gradient Descent**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_regression
```

```python
# Making Dataset
X,y=make_regression(n_samples=50,n_features=1,noise=30,random_state=42)
X = (X - X.mean()) / X.std()
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(X,y)
print("slope: ",lr.coef_)
print("Intercept: ",lr.intercept_)
plt.plot(X,lr.predict(X),color='red')
plt.scatter(X,y)
plt.show()
```

```python
import numpy as np

class GDRegressor:
    def __init__(self, learning_rate, epochs):
        self.lr = learning_rate
        self.epochs = epochs
        self.slope = 20
        self.intercept = 6

    def fit(self, X_train, y_train):
        for i in range(self.epochs):
            y_pred = self.intercept + self.slope * X_train

            slope_change = np.mean((y_pred - y_train) * X_train)
            intercept_change = np.mean(y_pred - y_train)

            self.slope = self.slope - self.lr * slope_change
            self.intercept = self.intercept - self.lr * intercept_change
```

```python
gr=GDRegressor(0.003,50)
gr.fit(X,y)
print("slope: ",gr.slope)
print("Intercept: ",gr.intercept)
plt.scatter(X,y)
plt.plot(X,gr.intercept+gr.slope*X,color='red')
plt.plot(X,lr.predict(X),color='green')
plt.show()
```
